```python
from optimal.RobotMujocoModel import RobotMujocoModel
from optimal.configuration import UR10, UR10_sandbox
import numpy as np

type = UR10
robot_mj_model = RobotMujocoModel(type.mjcf_path, type.list_of_joints, start_position=type.start_position)
print(f"joint names: {robot_mj_model.joint_names}")
print(f"q_indices: {robot_mj_model.q_indices}")
print(f"v_indices: {robot_mj_model.v_indices}")
print(f"u_indices: {robot_mj_model.u_indices}")
robot_mj_model.snapshot()
print(f"current joint positions: {robot_mj_model.current_joint_positions()}")
print(f"current joint velocities: {robot_mj_model.current_joint_velocities()}")
robot_mj_model.set_joint_positions(np.array([0.0, -np.pi/4, np.pi/2, -np.pi/4, -np.pi/2, 0.0]))
robot_mj_model.snapshot()
print(f"current joint positions: {robot_mj_model.current_joint_positions()}")
print(f"current joint velocities: {robot_mj_model.current_joint_velocities()}")
robot_mj_model.reset()
robot_mj_model.snapshot()
print(f"current joint positions: {robot_mj_model.current_joint_positions()}")
print(f"current joint velocities: {robot_mj_model.current_joint_velocities()}")
```
