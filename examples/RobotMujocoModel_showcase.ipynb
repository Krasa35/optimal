#   RobotMujocoModel
This class encapsulates the loading and interaction with a MuJoCo robot model. Below are use cases and examples of how to use the `RobotMujocoModel` class.

```python
from pathlib import Path
from optimal.RobotMujocoModel import RobotMujocoModel

MODEL_PATH = Path.cwd().parent / "models" / "UR5gripper_2_finger_KKI.xml"
MujModel = RobotMujocoModel(MODEL_PATH)
print(f"MujModel.joint_names: {MujModel.joint_names}")
print(f"MujModel.actuated_joint_names: {MujModel.actuated_joint_names}")
```

```python
import matplotlib.pyplot as plt

START_JOINT_POSITIONS = {
    "shoulder_pan_joint": 0.0,
    "shoulder_lift_joint": -1.57,
    "elbow_joint": 1.57,
    "wrist_1_joint": 0.0,
    "wrist_2_joint": 0.0,
    "wrist_3_joint": 0.0,
    "base_to_lik": 0.0,
}

MujModel.reset()
print(f"Joint values after initialization: {MujModel.joint_positions}")
plt.imshow(MujModel.snapshot())
plt.axis('off')
plt.show()
MujModel.joint_positions = START_JOINT_POSITIONS
print(f"Joint values after setting to START_JOINT_POSITIONS: {MujModel.joint_positions}")
plt.imshow(MujModel.snapshot())
plt.axis('off')
plt.show()
MujModel.joint_positions = {"shoulder_pan_joint": -0.5}
print(f"Joint values after setting shoulder_pan_joint to -0.5: {MujModel.joint_positions}")
plt.imshow(MujModel.snapshot())
plt.axis('off')
plt.show()
MujModel.joint_positions = MujModel.random_actuated_q()
print(f"Joint values after setting to random actuated q: {MujModel.joint_positions}")
plt.imshow(MujModel.snapshot())
plt.axis('off')
plt.show()
```

```python
print(f"Indices of positions: {MujModel.q_indices()}")
print(f"Indices of velocities: {MujModel.v_indices()}")
print(f"Indices of velocities: {MujModel.joint_names}")
print(f"Indices of velocities: {MujModel.actuated_joint_names}")
print(f"Indices of position of joint shoulder_pan_joint: {MujModel.q_indices(['shoulder_pan_joint'])}")
```
